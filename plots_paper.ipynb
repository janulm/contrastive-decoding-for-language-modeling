```python
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter
import numpy as np

# -------- 1) Data (edit here) -----------------------------------------------
# Groups: Vanilla (no extra sampling), then Top-p 97/95/90, Top-k 50/100/200
categories = ['Vanilla','P-97','P-95','P-90','K-50','K-100','K-200']

# Perplexity: relative improvement vs baseline (%)
ppl_no_contrast = [3.68, 3.78, 2.93, 2.27, 2.36, 2.66, 3.29]
ppl_contrast    = [2.98, 2.46, 2.69, 2.93, 2.30, 2.73, 2.80]

# MRPvB: relative improvement vs baseline (%)
mrp_no_contrast = [2.96, 2.30, 2.33, 2.48, 3.48, 2.49, 3.65]
mrp_contrast    = [4.90, 2.98, 4.54, 4.91, 4.64, 4.90, 5.69]

# -------- 2) Compact style controls -----------------------------------------
plt.rcParams['font.size'] = 9.5
# Use Times New Roman everywhere (Times-like math via STIX)
plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman", "Times", "Nimbus Roman No9 L", "STIXGeneral"],
    "mathtext.fontset": "stix",  # makes math look like Times
    "pdf.fonttype": 42,          # embed TrueType in PDFs
    "ps.fonttype": 42,           # embed TrueType in PS
})
color_nc = '#4C78A8'   # blue
color_cd = '#F58518'   # orange
hatch_placeholder = '//'

# --- SPACING KNOBS -----------------------------------------------------------
# Controls inside a group (between the two bars)
center_gap  = 0.00     # free space between the two bars in the same group (axis x-units)
bar_width   = 0.22     # width of each bar

# Controls between groups
group_gap   = 0.15     # extra whitespace between adjacent groups (axis x-units)
# The distance between group centers is:
group_step  = 2*bar_width + center_gap + group_gap

# Optional manual y-limits (set to None for auto with padding)
y1_limits = None
y2_limits = None
auto_pad  = 0.35

# -------- 3) Helpers ---------------------------------------------------------
def add_grouped_bars(ax, x, vals_a, vals_b, label_a, label_b, color_a, color_b,
                     bar_width, center_gap):
    """Place two bars per group centered around x, with a controllable gap."""
    lefts  = x - (bar_width/2 + center_gap/2)
    rights = x + (bar_width/2 + center_gap/2)

    bars_a = ax.bar(lefts, [v if v is not None else 0 for v in vals_a],
                    width=bar_width, color=color_a, label=label_a)

    bars_b = []
    for xi, v in zip(rights, vals_b):
        if v is None:
            b = ax.bar(xi, 0.01, width=bar_width, color='none',
                       edgecolor='gray', hatch=hatch_placeholder, alpha=0.9)
            ax.text(xi, 0.06, 'N/A', ha='center', va='bottom', fontsize=11, color='gray', rotation=90)
            bars_b.append(b[0])
        else:
            b = ax.bar(xi, v, width=bar_width, color=color_b, label=label_b)
            bars_b.append(b[0])
    return bars_a, bars_b

def apply_limits(ax, data_a, data_b, limits):
    if limits is not None:
        ax.set_ylim(*limits)
    else:
        vals = [v for v in (list(data_a) + list(data_b)) if v is not None]
        ymin, ymax = min(vals), max(vals)
        ax.set_ylim(max(0, ymin - auto_pad), ymax + auto_pad)

def add_vanilla_lines(ax, vanilla_nc, vanilla_cd):
    ls_dots = (0, (1, 2))  # on=1pt, off=2pt
    ax.axhline(vanilla_nc, color=color_nc, linewidth=1.1, linestyle=ls_dots, zorder=5)
    ax.axhline(vanilla_cd, color=color_cd, linewidth=1.1, linestyle=ls_dots, zorder=5)

# -------- 4) Build compact figure -------------------------------------------
x = np.arange(len(categories)) * group_step  # group centers spaced by group_step
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(5.0, 4.4), sharex=True)

# Top: Perplexity improvement
ba1, bb1 = add_grouped_bars(ax1, x, ppl_no_contrast, ppl_contrast,
                            'No-Contrast', 'CD-Early-500 (Contrast)', color_nc, color_cd,
                            bar_width, center_gap)
ax1.set_ylabel('Perplexity Δ vs. BASELINE (%)')
ax1.yaxis.set_major_formatter(PercentFormatter(xmax=100, decimals=1))
ax1.axhline(0, color='gray', linewidth=0.8)
ax1.grid(True, axis='y', linewidth=0.5, alpha=0.5)
apply_limits(ax1, ppl_no_contrast, ppl_contrast, y1_limits)

# Add Vanilla reference lines (Perplexity)
vanilla_idx = categories.index('Vanilla')
add_vanilla_lines(ax1, ppl_no_contrast[vanilla_idx], ppl_contrast[vanilla_idx])

# Bottom: MRPvB improvement
ba2, bb2 = add_grouped_bars(ax2, x, mrp_no_contrast, mrp_contrast,
                            'No-Contrast', 'CD-Early-500 (Contrast)', color_nc, color_cd,
                            bar_width, center_gap)
ax2.set_ylabel(r'$\mu_{\Delta \mathrm{REL}}$ vs BASELINE (%)')
ax2.set_xlabel('Sampling setting')
ax2.yaxis.set_major_formatter(PercentFormatter(xmax=100, decimals=1))
ax2.axhline(0, color='gray', linewidth=0.8)
ax2.grid(True, axis='y', linewidth=0.5, alpha=0.5)
apply_limits(ax2, mrp_no_contrast, mrp_contrast, y2_limits)

# Add Vanilla reference lines (MRPvB)
add_vanilla_lines(ax2, mrp_no_contrast[vanilla_idx], mrp_contrast[vanilla_idx])

# X ticks / labels: short, slightly rotated to save space
ax2.set_xticks(x)
ax2.set_xticklabels(categories, rotation=28, ha='right', rotation_mode='anchor')

# Tight margins & subtle subgroup divider between P and K
for ax in (ax1, ax2):
    ax.margins(x=0.01)
divider_x = 3.5 * group_step  # between P-90 (index 3) and K-50 (index 4)
for ax in (ax1, ax2):
    ax.axvline(divider_x, color='lightgray', linestyle=':', linewidth=0.7)

# Compact legend at the bottom
fig.subplots_adjust(hspace=0.08, bottom=0.20, left=0.23, right=0.98, top=0.98)
fig.legend([ba1[0], bb1[0]], ['No-Contrast', 'CD-Early-500 (Contrast)'],
           loc='lower center', ncol=2, frameon=False, fontsize=10)

plt.show()

# Optional saves
fig.savefig('plots/compact_grouped_bars_vanilla_first.pdf', bbox_inches='tight')
fig.savefig('plots/compact_grouped_bars_vanilla_first.png', dpi=300, bbox_inches='tight')
```
